## My solution for problem 1

```python
# Overview of the function.
x = np.linspace(-10, 10, 2000)
y = mx.lab_2_problem_1(x, student_id)
plt.figure(figsize=(8, 5))
plt.plot(x, y, label="f(x)")
plt.axhline(0, color="black", linewidth=0.8)
plt.title("Problem 1: Function overview")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.xlim(-10, 10)
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()

# Separate close-ups of each crossing of y = 0.
# Use new, closely spaced x values to keep each zoom smooth.
root_windows = [(-2.19, -2.15), (-0.98, -0.94),
                (3.97, 4.01), (5.14, 5.18)]
for number, (left, right) in enumerate(root_windows, start=1):
    x = np.linspace(left, right, 1000)
    y = mx.lab_2_problem_1(x, student_id)
    plt.figure(figsize=(8, 5))
    plt.plot(x, y, label="f(x)")
    plt.axhline(0, color="black", linewidth=0.8, label="y = 0")
    plt.xlim(left, right)
    plt.ylim(-1.5, 1.5)
    plt.xticks(np.linspace(left, right, 9))
    plt.title(f"Problem 1: Root {number} (zoomed)")
    plt.xlabel("x")
    plt.ylabel("f(x)")
    plt.grid(True)
    plt.legend()
    plt.tight_layout()
    plt.show()
```

### Graphical results
The four close-ups show crossings of the horizontal line $y=0$. Reading the x-coordinates to two decimal places gives:

$x \approx -2.17,\ -0.96,\ 3.99,\ 5.16$.

### My solution

```python
# Personalized functions displayed above for student ID 2622764.
x = np.linspace(0, 10, 2000)
f = np.sqrt(3.0 * x)
g = (x - 5.82)**2
plt.figure(figsize=(8, 5))
plt.plot(x, f, label="f(x) = sqrt(3.0x)")
plt.plot(x, g, label="g(x) = (x - 5.82)^2")
plt.title("Problem 2: Function overview")
plt.xlabel("x")
plt.ylabel("y")
plt.xlim(0, 10)
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()

# One separate close-up for each intersection.
intersection_windows = [(3.945, 3.985, 3.40, 3.50),
                        (8.015, 8.055, 4.85, 4.97)]
for number, (left, right, bottom, top) in enumerate(intersection_windows, start=1):
    x = np.linspace(left, right, 1000)
    f = np.sqrt(3.0 * x)
    g = (x - 5.82)**2
    plt.figure(figsize=(8, 5))
    plt.plot(x, f, label="f(x) = sqrt(3.0x)")
    plt.plot(x, g, label="g(x) = (x - 5.82)^2")
    plt.xlim(left, right)
    plt.ylim(bottom, top)
    plt.xticks(np.linspace(left, right, 9))
    plt.yticks(np.arange(bottom, top + 0.001, 0.01))
    plt.title(f"Problem 2: Intersection {number} (zoomed)")
    plt.xlabel("x")
    plt.ylabel("y")
    plt.grid(True)
    plt.legend()
    plt.tight_layout()
    plt.show()
```

### Graphical results
Reading the two close-ups, the intersection points are approximately **(3.96, 3.45)** and **(8.04, 4.91)**. At each point, both functions have the same y-value. These are graphical estimates rounded to two decimal places.
