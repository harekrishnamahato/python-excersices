### My solution

```python
# Step size 0.005, including both endpoints.
x = np.linspace(-10, 10, 4001)

# Each boolean mask selects one region of the function.
y = ((x < -0.5) * 0
     + ((x >= -0.5) & (x < 2.5)) * (0.5 + x)
     + ((x >= 2.5) & (x < 5.5)) * (5.5 - x)
     + (x >= 5.5) * 0)

plt.figure(figsize=(8, 5))
plt.plot(x, y, label="f(x)")
plt.xlim(-10, 10)
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Problem 1: Triangular piecewise function")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

The triangle starts at **(-0.5, 0)**, reaches its peak at **(2.5, 3)**, and ends at **(5.5, 0)**. It is zero outside this interval. The pieces agree at all three boundaries.

```python
# Run the code below to see your problem text
txt = mx.lab_3_problem_2(student_id)
display(Markdown(txt))
```

### My solution

```python
# Use the same dense x-array from Problem 1.
y = ((x < -5.2) * 286.73
     + ((x >= -5.2) & (x < 4.1))
       * (x**3 - 7.11*x**2 - 50.39*x + 357.56)
     + (x >= 4.1) * 100.36)

plt.figure(figsize=(8, 5))
plt.plot(x, y, label="f(x)")
plt.xlim(-10, 10)
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Problem 2: Piecewise cubic function")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

The function is constant at **286.73** to the left of **-5.2**, follows the cubic between **-5.2** and **4.1**, and is constant at **100.36** from **4.1** onward. The masks cover every x-value without overlap.

The displayed constants are rounded to two decimals: the cubic gives 286.7256 at -5.2 and 100.3609 at 4.1. The tiny rounding differences are retained to match the personalized definition exactly.
