```python
import matplotlib.pyplot as plt 
import numpy as np
```

```python


x = np.linspace(0 , 4)
y = 0.1 * x ** 4 - 0.5 * x * 3 + 0.2  * x** 2 - 0.1 * x+ 4

fig, ax = plt.subplots( figsize  = (10,6))

plt.plot(x , y, linestyle = "--", linewidth = 2.5, label = "4th polynomilal function")
plt.grid(True)
plt.xlabel("X")
plt.ylabel("Y")
plt.title("the graph of the function", fontsize = 19)
plt.legend()
plt.show
```

```python
#working with multiple-line plot

fig, ax = plt.subplots()

x = np.linspace(0,4)
y1 = x - 2
y2 = x**2 - 5*x

plt.plot(x, y1, label = "$ y  = x - 2 $")
plt.plot(x, y2, label = "$ y  = x^2 - 4x $")
plt.grid(True)
plt.xlabel("X")
plt.ylabel("Y")
plt.title("the graph of the function", fontsize = 23)
plt.legend()
plt.show
```

```python
#work with multiple subplots 

fig, ax = plt.subplots(2, figsize = (10,8))

x = np.linspace(0, 5)

y1 = x - 2
plt.sca(ax[0])
plt.plot(x, y1,  label = "$ y  = x - 2 $")
plt.grid(True)
plt.xlabel("X")
plt.ylabel("Y")
plt.title("the graph of the function", fontsize = 23)
plt.legend()
plt.show
y2 = x**2 - 5*x
plt.sca(ax[1])
plt.plot(x, y2,  label = "$ y  = x - 2 $")
plt.grid(True)
plt.xlabel("X")
plt.ylabel("Y")
plt.title("the graph of the function", fontsize = 23)
plt.legend()
plt.show
```

```python
x = np.linspace(-5, 5, 200)
y = x**2 - 2*x - 5

plt.figure(figsize=(10,5))
plt.plot(x,y, label = "f(x)")
plt.axhline(0, color="g",  linestyle = "--", label = "y = 0")
plt.title ("graph solving math problem")
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.grid(True)
plt.show()
```
